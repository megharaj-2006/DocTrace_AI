```python
from PIL import Image
import imagehash

h1 = imagehash.phash(Image.open("img1.jpg"))
h2 = imagehash.phash(Image.open("img2.jpg"))
print("Hash 1:", h1)
print("Hash 2:", h2)
print("Distance:", h1 - h2)
```
